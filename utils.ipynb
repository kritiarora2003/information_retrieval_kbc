```python
import pandas as pd
import pickle

# 1️⃣ Load the embeddings DataFrame
with open("embeddings_output/embeddings_new_2_with_prop_afterprop.pkl", "rb") as f:
    emb_df = pickle.load(f)

# 2️⃣ Select only the needed columns + filename
cols_to_add = ['filename', 'propositions', 'propositions_en', 'propositions_en_set_list', 'embeddings_propositions_en']
emb_df = emb_df[cols_to_add]

# 3️⃣ Merge without converting to string
# Standardize filenames in both DataFrames
files['filename'] = files['filename'].astype(str).str.strip().str.zfill(6)
emb_df['filename'] = emb_df['filename'].astype(str).str.strip().str.zfill(6)

# Optional: remove duplicates in emb_df
emb_df = emb_df.drop_duplicates(subset='filename')

# Merge
files = files.merge(emb_df, on='filename', how='left')
files 
```

```python
files[files['query']==True]
```

```python
import pickle 
filename = 'embeddings_output/embeddings_new_2_with_prop_afterpropparasent.pkl'

with open(filename, 'wb') as f:
    pickle.dump(files, f)
print("Saved pickle: ", filename) 
```
